## Data preparation

```python
for col in df.columns[df.isna().any()].tolist():
    print(f"{col} : {df[col].dtype}")
```

```python
for col in df.columns[df.isna().any()].tolist():
    temptype = df[col].dtype
    if temptype == str:
        df[col] = df[col].fillna("NA")
    else:
        df[col] = df[col].fillna(0)

# note : sample of df was observed with actual values as well.
# Merely having int or float dtype does not ensure that the column represents numerical feature
```
