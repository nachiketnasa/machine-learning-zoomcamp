## Sum of weights

```python
res = df.loc[(df['origin'] == 'Asia')][['vehicle_weight', 'model_year']].head(7)
res
```

```python
x = res.to_numpy()
x
```

```python
x.T
```

```python
xtx = x.T.dot(x)
```

```python
xtx
```

```python
xtx_inv = np.linalg.inv(xtx)
xtx_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = xtx_inv.dot(x.T)
w
```

```python
res = w.dot(y)
res
```
